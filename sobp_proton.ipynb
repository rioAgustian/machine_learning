$\begin{bmatrix}
z_1 \\ z_2 \\ z_3 \\ ... \\ z_n \\  
\end{bmatrix}$
$=$
$ w_1 \begin{bmatrix}
y_{11} \\ y_{12} \\ y_{13} \\ ... \\ y_{1n} \\  
\end{bmatrix}$
$ + w_2 \begin{bmatrix}
y_{21} \\ y_{22} \\ y_{23} \\ ... \\ y_{2n} \\  
\end{bmatrix}$
$ + w_3 \begin{bmatrix}
y_{31} \\ y_{32} \\ y_{33} \\ ... \\ y_{3n} \\  
\end{bmatrix}$
$ ... + w_m \begin{bmatrix}
y_{m1} \\ y_{m2} \\ y_{m3} \\ ... \\ y_{mn} \\  
\end{bmatrix}$

```python
import numpy as np
import matplotlib.pyplot as plt

def bragg(x, xp, A):
    return A*np.exp(-((x-xp)/0.5)**2)

x = np.linspace(0, 20, 100)
y_data = []

k = 1
for i in np.arange(5, 15, 0.3):
    y_data.append(bragg(x, i, k))
    k = k - 0.01
    
for i in range(len(y_data)):
    plt.plot(x, y_data[i])

y = 0
for i in range((len(y_data))):
    y = y + y_data[i]
    
plt.plot(x, y)
```

```python
z = np.zeros(y.shape)
# z[20:76] = 2

y_array = np.array(y_data)
y_array.shape
```

```python
y_array.reshape(100, 34)
z.reshape(100)

sol = np.linalg.lstsq(y_array.T, z, rcond=-1)

sol[0]
```
